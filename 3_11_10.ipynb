```python
from bitcoinlib.transactions import Transaction
from bitcoinlib.keys import Key

sender_key = Key()
receiver_key = Key()

print("Sender Private Key:")
print(sender_key.wif())

print("\nSender Address:")
print(sender_key.address())

print("\nReceiver Address:")
print(receiver_key.address())

previous_tx = Transaction()

# Give the sender an output of 200,000 satoshis
previous_tx.add_output(
    value=200,
    address=sender_key.address()
)

print("\nPrevious transaction created.")

previous_txid = previous_tx.txid

print("\nPrevious Transaction ID:")
print(previous_txid)

tx = Transaction()

print("\nNew transaction created.")

tx.add_input(
    prev_txid=previous_txid,
    output_n=0
)

print("Input added.")

# Add output to send money to receiver
tx.add_output(
    value=100000,
    address=receiver_key.address()
)

print("Output added.")

# ---------------------------------------------------------
# Sign the transaction
# ---------------------------------------------------------
tx.sign(sender_key)

print("Transaction signed successfully.")


# ---------------------------------------------------------
# MONEY FLOW / BALANCE CHECK
# ---------------------------------------------------------

initial_balance = previous_tx.outputs[0].value
amount_sent = tx.outputs[0].value

print("\n" + "=" * 50)
print("MONEY FLOW SUMMARY")
print("=" * 50)

print(f"\n💰 Sender's UTXO:       {initial_balance:,} satoshis")
print(f"📤 Requested Amount:    {amount_sent:,} satoshis")


# ---------------------------------------------------------
# Check whether sender has enough funds
# ---------------------------------------------------------

if amount_sent > initial_balance:

    print("\n❌ TRANSACTION INVALID")
    print("❌ Insufficient funds!")
    print(
        f"Sender has {initial_balance:,} satoshis "
        f"but tried to send {amount_sent:,} satoshis."
    )

    print("\nTransaction rejected.")
    print("No money is transferred.")

else:

    # Remaining amount becomes change
    remaining_balance = initial_balance - amount_sent

    print(f"💵 Remaining Balance:   {remaining_balance:,} satoshis")

    print(
        f"\nVerification: "
        f"{initial_balance:,} = "
        f"{amount_sent:,} + "
        f"{remaining_balance:,}"
    )

    print("\n✅ Sufficient funds.")
    print("✅ Transaction can be accepted.")

    print("\nSigned Transaction HEX:")
    print(tx.raw_hex())
```

```python
signature_valid = tx.verify()

if tx.verify():
    print("Signature is VALID")
else:
    print("Signature is INVALID")
```

```python
tx.outputs[0].value = 150000

signature_valid = tx.verify()

if tx.verify():
    print("Signature is VALID")
else:
    print("Signature is INVALID")
```

```python
expected_address = receiver_key.address()

actual_address = tx.outputs[0].address

print("Expected Receiver Address:", expected_address)
print("Actual Receiver Address:  ", actual_address)

address_valid = (actual_address == expected_address)
if address_valid:
    print("Receiver address is VALID")
else:
    print("Receiver address is INVALID")
```

```python
if signature_valid and address_valid:
    print("\n✓ Transaction verification SUCCESS")
else:
    print("\n✗ Transaction verification FAILED")
```
